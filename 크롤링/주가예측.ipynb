```python
import pandas as pd
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from datetime import date, datetime
import pytz
```

```python
seoul_tz = pytz.timezone("Asia/Seoul")
today = datetime.now(seoul_tz).strftime("%Y-%m-%d")
today
# CSV 파일 읽기
df = pd.read_csv(f"/Users/kjb/Desktop/hateslop/프로젝트/테스트데이터/({today})테스트.csv")
df
```

```python
features = [
    'RSI', 'BB_Dist', 'MACD_6_13_4', 'MACDh_6_13_4', 'MACDs_6_13_4',
    'STOCHk_5_3_1', 'STOCHd_5_3_1', 'OBV','ATR',
    'Golden_Cross','ADX', 'CCI', 'ROC', 'CMF','ADL'
]

test = df[features]
test
```

```python
from xgboost import XGBClassifier

# 모델 로드
model = XGBClassifier()
model.load_model("/Users/kjb/Desktop/hateslop/프로젝트/예측모델/xgb_model.json")

# 예측 사용 가능
y_pred = model.predict(test)
y_proba = model.predict_proba(test)
```

```python
import pandas as pd

# 위키피디아의 S&P 500 구성 종목 목록 URL
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

# HTML 테이블을 읽어오기
tables = pd.read_html(url)
df = tables[0]  # 첫 번째 테이블이 S&P 500 구성 종목 목록


# 티커와 기업명 추출
ticker_Group= df['Symbol'].tolist()
company_names = df['Security'].tolist()

ticker_Group = [t.replace(".", "-") for t in ticker_Group]
company_names = [t.replace(".", "-") for t in company_names]
ticker_names = dict(zip(ticker_Group, company_names))

ticker_Group
ticker_names
for i, ticker in enumerate(ticker_Group):
    kor_name = ticker_names.get(ticker, ticker)
    prob = y_proba[i][1] * 100
    direction = "상승" if prob >= 50 else "하락"
    print(f"{kor_name} 상승확률 : {prob:.2f}%, {kor_name} 상승여부 : {direction}")
```

```python
import pandas as pd

results = []

for i, ticker in enumerate(ticker_Group):
    kor_name = ticker_names.get(ticker, ticker)
    pred = y_pred[i]
    prob = y_proba[i][1] * 100
    results.append({
        "Ticker": ticker,
        "종목": kor_name,
        "상승확률": round(prob, 2),
        "상승여부": "상승" if prob > 50 else "하락"
    })

df_result = pd.DataFrame(results)
df_result = df_result.iloc[:,1:4]
df_result
```

```python
import os
save_path = f"/Users/kjb/Desktop/hateslop/프로젝트/예측데이터/({today})예측.csv"
df_result.to_csv(save_path, index=False)
```
